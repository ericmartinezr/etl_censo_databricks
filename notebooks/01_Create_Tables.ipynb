# Creación de tablas

# Crea DB

```python
%sql
CREATE DATABASE IF NOT EXISTS bronze;
CREATE DATABASE IF NOT EXISTS silver;
CREATE DATABASE IF NOT EXISTS gold;
```

# Crea tablas y las trunca

## Bronze

```python
%sql
DROP TABLE IF EXISTS bronze.viviendas
```

```python
%sql
CREATE TABLE IF NOT EXISTS bronze.viviendas (
    id_vivienda STRING,
    region STRING,
    provincia STRING,
    comuna STRING,
    comuna_bajo_umbral STRING,
    area STRING,
    tipo_operativo STRING,
    cant_hog STRING,
    cant_per STRING,
    p2_tipo_vivienda STRING,
    p3a_estado_ocupacion STRING,
    p3b_estado_ocupacion STRING,
    p4a_mat_paredes STRING,
    p4b_mat_techo STRING,
    p4c_mat_piso STRING,
    p5_num_dormitorios STRING,
    p6_fuente_agua STRING,
    p7_distrib_agua STRING,
    p8_serv_hig STRING,
    p9_fuente_elect STRING,
    p10_basura STRING,
    p11a_num_personas STRING,
    p11b_comparte_gasto STRING,
    p11c_num_hogar STRING,
    indice_hacinamiento INT
)
USING DELTA
LOCATION 'abfss://bronze@censostorageaccount.dfs.core.windows.net'
CLUSTER BY (id_vivienda);
```
